## MgO Array Creation

```python
from pymatgen.core.lattice import Lattice

def generate_mgo_coordinates(a, b, c, Mg_element="Mg", O_element="O", 
                            dx=0.0, dy=0.0, dz=0.0, z_shift=0.0):
    """
    Generates atomic coordinates for an MgO conventional unit cell (rock-salt structure, Fm-3m).
    
    Args:
        a, b, c: Lattice parameters (Å)
        Mg_element: Element for Mg sites (default: "Mg")
        O_element: Element for O sites (default: "O")
        dx, dy, dz: Fractional coordinate shifts
        z_shift: Additional z-axis shift in Å (real coordinates)
        
    Returns:
        Dictionary containing:
        - 'Mg_atoms': (element, coords) for Mg sites (4a Wyckoff positions)
        - 'O_atoms': (element, coords) for O sites (4b Wyckoff positions)
        - 'last_z_coordinate': z-coordinate of the last atom in Cartesian coordinates
    """
    # Define fractional coordinates for MgO conventional unit cell (space group Fm-3m)
    # 4a sites (Mg): (0,0,0), (0.5,0.5,0), (0.5,0,0.5), (0,0.5,0.5)
    # 4b sites (O): (0.5,0.5,0.5), (0,0,0.5), (0,0.5,0), (0.5,0,0)
    frac_coords = [
        # 4a sites (Mg atoms)
        [0.00, 0.00, 0.00],  # Mg
        [0.50, 0.50, 0.00],  # Mg
        [0.50, 0.00, 0.50],  # Mg
        [0.00, 0.50, 0.50],  # Mg
        # 4b sites (O atoms)
        [0.50, 0.50, 0.50],  # O
        [0.00, 0.00, 0.50],  # O
        [0.00, 0.50, 0.00],  # O
        [0.50, 0.00, 0.00],  # O
    ]
    
    # Assign species to each site
    species = [
        Mg_element, Mg_element, Mg_element, Mg_element,  # 4a sites
        O_element, O_element, O_element, O_element        # 4b sites
    ]
    
    # Apply fractional shifts with wrapping to [0,1)
    shifted_frac_coords = [
        [(x + dx) % 1, (y + dy) % 1, (z + dz) % 1]
        for x, y, z in frac_coords
    ]
    
    # Convert to Cartesian coordinates
    lattice = Lattice.from_parameters(a, b, c, 90, 90, 90)  # Cubic lattice for MgO
    cart_coords = [lattice.get_cartesian_coords(coord) for coord in shifted_frac_coords]
    
    # Apply z_shift in Cartesian coordinates (Å)
    if z_shift != 0:
        cart_coords = [[x, y, z + z_shift] for x, y, z in cart_coords]
    
    # Group coordinates by atom type
    return {
        'Mg_atoms': (Mg_element, cart_coords[0:4]),  # 4a sites
        'O_atoms': (O_element, cart_coords[4:8]),    # 4b sites
        'last_z_coordinate': cart_coords[-1][2]      # Last z-coordinate
    }
```

### Provide the number of layers. 

```python
# ===================================== Enter the Number of Layers ===========================================
n_layers = 4  # Example input
# ============================================================================================================
```

```python
import math
from pymatgen.core import Structure, Lattice
from crystal_toolkit.components import StructureMoleculeComponent
from dash import Dash, html
import random

# Assume generate_mgo_coordinates function exists as referenced in the original code
# If not defined, it should generate MgO coordinates with the specified parameters

layers_per_unit_cell = 2
n_cells = math.ceil(n_layers / layers_per_unit_cell)

# Lattice parameters (cubic for MgO)
a, b, c = 4.03, 4.03, 4.03 * n_cells  # Adjust c for total length of cascaded cells

# Generate n cascaded MgO unit cells
groups = []
for i in range(n_cells):
    # z_shift for each group is c * i (c = 4.03 for MgO)
    z_shift = 4.03 * i
    group = generate_mgo_coordinates(a, b, c=4.03, dx=0, dy=0.0, dz=0.0, z_shift=z_shift)
    groups.append(group)

# Combine species and coordinates from all groups
MgO_species = []
MgO_coords = []
MgO_magmoms = []
for group in groups:
    # Add Mg atoms (4 positions)
    MgO_species.extend([group['Mg_atoms'][0]] * 4)
    MgO_coords.extend(group['Mg_atoms'][1])
    MgO_magmoms.extend([0.0] * 4) 
    
    # Add O atoms (4 positions)
    MgO_species.extend([group['O_atoms'][0]] * 4)
    MgO_coords.extend(group['O_atoms'][1])
    MgO_magmoms.extend([0.0] * 4) 

# Create pymatgen Structure
lattice = Lattice.from_parameters(a, b, c, 90, 90, 90)
structure = Structure(
    lattice=lattice,
    species=MgO_species,
    coords=MgO_coords,
    coords_are_cartesian=True,
    site_properties={"magmom": MgO_magmoms}
)

# Visualize the structure
try:
    scene_settings = {
        "polyhedra_opacity": 0.1,
        "unit_cell_opacity": 0.05,
    }
    component = StructureMoleculeComponent(structure, scene_settings=scene_settings)
    
    app = Dash(__name__)
    app.layout = html.Div([
        html.H2("MgO Cascaded Structure Visualization"),
        component.layout(),
        html.H3(f"Number of unit cells: {n_cells}"),
        html.P(f"z-coordinate of the last atom in the cascade: {groups[-1]['last_z_coordinate']:.8f} Å")
    ])
    
    # Print the results
    print(f"MgO_species (n={n_cells}) =", MgO_species)
    print(f"\nMgO_coords (n={n_cells}) = [")
    for coord in MgO_coords:
        print(f"    {coord},")
    print("]")
    print(f"\nz-coordinate of the last atom in the cascade = {groups[-1]['last_z_coordinate']:.8f}")

    # Run Dash app
    app.run(debug=False, port=random.randint(8051, 8999))
except ImportError:
    print("\nNote: crystal-toolkit not available for visualization")
    # Print the results without visualization
    print(f"MgO_species (n={n_cells}) =", MgO_species)
    print(f"\nMgO_coords (n={n_cells}) = [")
    for coord in MgO_coords:
        print(f"    {coord},")
    print("]")
    print(f"\nz-coordinate of the last atom in the cascade = {groups[-1]['last_z_coordinate']:.8f}")
```

```python
# Extract species, coordinates, and magnetic moments from the structure
MgO_species = [site.species_string for site in structure]
MgO_coords = [site.coords.tolist() for site in structure]
MgO_magmoms = structure.site_properties['magmom']

# Now you can use these lists to create a new structure or for other purposes
print("Species:", MgO_species)
print("Coordinates:", MgO_coords)
print("Magnetic moments:", MgO_magmoms)
```
