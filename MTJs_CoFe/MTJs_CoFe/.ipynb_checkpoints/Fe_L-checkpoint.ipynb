## Fe Electrode_L

```python
from pymatgen.core.lattice import Lattice
import math

def generate_fe_coordinates(a, b, c, Fe_element="Fe", 
                             dx=0.0, dy=0.0, dz=0.0, z_shift=0):
    """
    Generates atomic coordinates for a simple Fe unit cell (custom 4-atom basis).
    
    Args:
        a, b, c: Lattice parameters (Å)
        Fe_element: Element for Fe sites (default: "Fe")
        dx, dy, dz: Fractional coordinate shifts
        z_shift: Additional z-axis shift in Å (real coordinates)
        
    Returns:
        Dictionary containing:
        - 'Fe_atoms': (element, coords) for Fe sites
        - 'last_z_coordinate': z-coordinate of the last atom in Cartesian coordinates
    """
    # Fractional coordinates of Fe atoms (from your script)
    frac_coords = [
        [0.50, 0.00, 0.5],
        [0.50, 0.50, 0.0],
        [0.0, 0.00, 0.0],
        [0.0, 0.50, 0.5],
    ]
    
    # Apply fractional shifts
    shifted_frac_coords = [
        [(x + dx) % 1, (y + dy) % 1, (z + dz) % 1]
        for x, y, z in frac_coords
    ]
    
    # Convert to Cartesian coordinates
    lattice = Lattice.from_parameters(a, b, c, 90, 90, 90)
    cart_coords = [lattice.get_cartesian_coords(coord) for coord in shifted_frac_coords]
    
    # Apply z_shift in Cartesian coordinates
    if z_shift != 0:
        cart_coords = [[x, y, z + z_shift] for x, y, z in cart_coords]
    
    return {
        'Fe_atoms': (Fe_element, cart_coords),
        'last_z_coordinate': cart_coords[-1][2]
    }
```

### Provide the Number of Layers

```python
# ===================================== Enter the Number of Layers ===========================================
n_layers = 5  # Example input
# ============================================================================================================a
```

```python
from pymatgen.core import Structure, Lattice
import numpy as np


layers_per_unit_cell = 2
n_cells = math.ceil(n_layers / layers_per_unit_cell)

# Lattice parameters (cubic for Fe)
a, b, c = 4.045, 4.045, 2.8605
magmom = 2.2  # Magnetic moment for all Fe atoms (in μB)

# Generate n cascaded Fe unit cells
groups = []
for i in range(n_cells):
    z_shift = c * i #+ 0.25*c  # z shift in angstroms
    group = generate_fe_coordinates(a, b, c, dx=0.5, dy=0.0, dz=0, z_shift=z_shift)
    groups.append(group)

# Combine species, coordinates and magnetic moments from all groups
Fe_species = []
Fe_coords = []
Fe_magmoms = []

for group in groups:
    # Add 4 Fe atoms with magnetic moments
    Fe_species.extend([group['Fe_atoms'][0]] * 4)
    Fe_coords.extend(group['Fe_atoms'][1])
    Fe_magmoms.extend([magmom] * 4)  # Add magmom for each atom

# Print the results
print(f"Fe_species (n={n_cells}) =", Fe_species)
print(f"\nFe_coords (n={n_cells}) = [")
for coord in Fe_coords:
    print(f"    {coord},")
print("]")
print(f"\nFe_magmoms (n={n_cells}) =", Fe_magmoms)
print(f"\nFinal z-coordinate in cascade = {groups[-1]['last_z_coordinate']:.4f} Å")
```

```python
from pymatgen.core import Structure, Lattice
import numpy as np
import math  # Added math import

class AtomData:
    def __init__(self, x: float, y: float, z: float, magmom: float, species: str):
        self.x = x
        self.y = y
        self.z = z
        self.magmom = magmom
        self.species = species
    
    def __repr__(self):
        return f"AtomData(x={self.x:.8f}, y={self.y:.8f}, z={self.z:.8f}, magmom={self.magmom:.3f}, species='{self.species}')"

class AtomicStructure:
    def __init__(self):
        self.atoms = []
    
    def add_atom(self, x: float, y: float, z: float, magmom: float, species: str):
        self.atoms.append(AtomData(x, y, z, magmom, species))
    
    def process_data(self, species_list, coord_list, magmom_list):
        for species, coord, magmom in zip(species_list, coord_list, magmom_list):
            x = float(coord[0])
            y = float(coord[1])
            z = float(coord[2])
            mag = float(magmom)
            self.add_atom(x, y, z, mag, species)
        
        # Sort atoms by z-coordinate
        self.atoms.sort(key=lambda atom: atom.z)
    
    def remove_last_n_atoms(self, n):
        """Remove the last n atoms (with highest z-coordinates)"""
        if len(self.atoms) >= n:
            self.atoms = self.atoms[:-n]
    
    def print_formatted(self):
        print("Index   X           Y           Z           Magmom      Species")
        print("-------------------------------------------------------------")
        for i, atom in enumerate(self.atoms, 1):
            print(f"{i:<6} {atom.x:<12.8f} {atom.y:<12.8f} {atom.z:<12.8f} {atom.magmom:<12.3f} {atom.species}")
    
    def create_pymatgen_structure(self, a: float, b: float, c: float):
        # Create lattice - now using max z-coordinate of remaining atoms
        if self.atoms:
            max_z = max(atom.z for atom in self.atoms)
            c = max_z + 0.5 * 2.85  # Buffer for visualization
        else:
            c = c  # Default value if no atoms
        
        lattice = Lattice.from_parameters(a, b, c, 90, 90, 90)
        
        # Prepare data for Structure
        species_list = [atom.species for atom in self.atoms]
        coords_list = [[atom.x, atom.y, atom.z] for atom in self.atoms]
        magmoms_list = [atom.magmom for atom in self.atoms]
        
        # Create Structure
        structure = Structure(
            lattice,
            species_list,
            coords_list,
            coords_are_cartesian=True,
            site_properties={"magmom": magmoms_list}
        )
        
        return lattice, structure

# Example usage
structure_data = AtomicStructure()
structure_data.process_data(Fe_species, Fe_coords, Fe_magmoms)

print("Original structure:")
structure_data.print_formatted()

# Remove atoms only if n_layers is not a multiple of 4
if n_layers % 2 != 0:
    atoms_to_remove = 2 * (math.ceil(n_layers / layers_per_unit_cell) * 2 - n_layers)
    structure_data.remove_last_n_atoms(atoms_to_remove)
    print(f"\nStructure after removing last {atoms_to_remove} atoms:")
    structure_data.print_formatted()
else:
    print("\nNo atoms removed (n_layers is a multiple of 4)")

# Create pymatgen Structure with remaining atoms
lattice, structure = structure_data.create_pymatgen_structure(a, b, c)

# Display lattice information
print("\nLattice parameters:")
print(f"a = {lattice.a:.6f} Å")
print(f"b = {lattice.b:.6f} Å")
print(f"c = {lattice.c:.6f} Å")
print(f"Volume = {lattice.volume:.6f} Å³")
print("\nLattice matrix:")
print(lattice.matrix)

# Visualize the structure (if crystal-toolkit is available)
try:
    from crystal_toolkit.components import StructureMoleculeComponent
    from dash import html, Dash
    import random
    
    scene_settings = {
        "polyhedra_opacity": 0.1,
        "unit_cell_opacity": 0.05,
    }
    component = StructureMoleculeComponent(structure, scene_settings=scene_settings)
    
    app = Dash(__name__)
    app.layout = html.Div([
        html.H2("Crystal Structure Visualization"),
        component.layout()
    ])
    
    app.run(debug=False, port=random.randint(8051, 8999))
except ImportError:
    print("\nNote: crystal-toolkit not available for visualization")
```
