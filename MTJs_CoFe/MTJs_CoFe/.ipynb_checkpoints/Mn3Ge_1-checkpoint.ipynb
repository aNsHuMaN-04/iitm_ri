## Extracting the Cell From the Materials Project
This is the example of the cell given in Fig 1(a) in the article by Stuart Parkin in 2017.

```python
"""
Parametric Heusler (inverse-type)
─────────────────────────────────────────────────────────────
• Change X_ELEMENT, Y_ELEMENT, Z_ELEMENT below to whatever atoms you like.
• Adjust a, c if you need different lattice parameters.
• Run the script and it will pop up a Dash page with:
    – 3-D viewer (Crystal-Toolkit)
    – raw CIF text of the generated structure
"""

from pymatgen.core import Lattice, Structure
from crystal_toolkit.components.structure import StructureMoleculeComponent
from dash import Dash, html
import random

# ── 1. USER CHOICES ─────────────────────────────────────────
X_ELEMENT = "Mn"   # the two symmetry-distinct X atoms (8 sites total)
Y_ELEMENT = "Al"   # Y atom (2 sites)
Z_ELEMENT = "Ga"   # Z atom (2 sites)

a = 4.06          # Å  ┐  tetragonal lattice parameters
c = 5.74          # Å  ┘  (set c = a if you want cubic)

# ── 2. BUILD LATTICE ────────────────────────────────────────
lattice = Lattice.from_parameters(a, a, c, 90, 90, 90)

# ── 3. FRACTIONAL COORDINATES (inverse Heusler, 2 f.u.) ────
#    ┌───────── top layer (z = 0 or ½) ─────────┐
#    │ Z @ (0,0,0);   Y @ (½,½,0)               │
#    │ Z @ (½,½,½);   Y @ (0,0,½)               │
#    └───────────────────────────────────────────┘
#    X atoms occupy the eight ¼,¾,±¼-type positions.
frac_coords = [
    # Z and Y framework
    [0.00, 0.00, 0.00],   # Z
    [0.50, 0.50, 0.00],   # Y
    [0.50, 0.50, 0.50],   # Z
    [0.00, 0.00, 0.50],   # Y, also can be called as X(II) 
    # Eight X positions (two symmetry-distinct sets)
    [0.50, 0.00, 0.25],   # All the X positons are X(I)
    [0.50, 0.00, 0.75],   
    [0.00, 0.50, 0.25],  
    [0.00, 0.50, 0.75],   
]

# Species list matching the coords order
species = [
    Z_ELEMENT, Y_ELEMENT, Z_ELEMENT, Y_ELEMENT,
    *([X_ELEMENT] * 4)
]

# ── 4. CREATE Structure ─────────────────────────────────────
structure = Structure(lattice, species, frac_coords)

# ── 5. CRYSTAL-TOOLKIT COMPONENT (with transparency) ────────
scene_settings = {
    "polyhedra_opacity": 0.1,
    "unit_cell_opacity": 0.05,
}
component = StructureMoleculeComponent(structure, scene_settings=scene_settings)

# ── 6. CIF EXPORT (rendered as <pre>) ───────────────────────
cif_string  = structure.to(fmt="cif")
cif_display = cif_string.replace("\n", "<br>").replace(" ", "&nbsp;")

# ── 7. DASH APP LAYOUT ──────────────────────────────────────
app = Dash(__name__)
app.layout = html.Div(
    [
        html.H2(f"🧱 Heusler Unit Cell: X = {X_ELEMENT}, "
                f"Y = {Y_ELEMENT}, Z = {Z_ELEMENT}"),
        component.layout(),
        html.H2("📄 Generated CIF"),
        html.Div(
            children=cif_display,
            style={
                "whiteSpace": "pre-wrap",
                "fontFamily": "monospace",
                "backgroundColor": "#f4f4f4",
                "padding": "10px",
                "borderRadius": "5px",
            },
        ),
    ]
)

# ── 8. RUN ──────────────────────────────────────────────────
app.run(debug=False, port=random.randint(8051, 8999))
```

```python
from pymatgen.core import Lattice, Structure
from crystal_toolkit.components.structure import StructureMoleculeComponent
from dash import Dash, html
import random

# ── 1. USER CHOICES ─────────────────────────────────────────
X_ELEMENT_2b = "Mn"  # Mn at 2b sites
X_ELEMENT_4d = "Mn"  # Mn at 4d sites
Z_ELEMENT = "Ge"     # Ge at 2a sites

a = 3.73             # Å  ┐  tetragonal lattice parameters
c = 7.26             # Å  ┘  (set c = a if you want cubic)

# ── 2. BUILD LATTICE ────────────────────────────────────────
lattice = Lattice.from_parameters(a, a, c, 90, 90, 90)

# ── 3. FRACTIONAL COORDINATES (body-centered tetragonal, 2 f.u.) ────
#    Wyckoff positions for space group I4/mmm (No. 139):
#    - 2a: (0, 0, 0), (0, 0, ½) for Ge
#    - 2b: (0, 0, ½), (0, 0, 0) for Mn (adjusted for 2b)
#    - 4d: (0, ½, ¼), (0, ½, ¾), (½, 0, ¼), (½, 0, ¾) for Mn
frac_coords = [
    # 2a sites (Ge)
    [0.00, 0.00, 0.00],  # Ge
    [0.50, 0.50, 0.50],  # Ge
    # 2b sites (Mn)
    [0.50, 0.50, 0.00],  # Mn
    [0.00, 0.00, 0.50],  # Mn
    # 4d sites (Mn)
    [0.00, 0.50, 0.25],  # Mn
    [0.00, 0.50, 0.75],  # Mn
    [0.50, 0.00, 0.25],  # Mn
    [0.50, 0.00, 0.75],  # Mn
]

# Species list matching the coords order
species = [
    Z_ELEMENT, Z_ELEMENT,  # 2a sites
    X_ELEMENT_2b, X_ELEMENT_2b,  # 2b sites
    *[X_ELEMENT_4d] * 4  # 4d sites
]

# ── 4. CREATE Structure with Magnetic Moments ───────────────
structure = Structure(lattice, species, frac_coords)
# Assign magnetic moments (Case 1: μ₂b = -3.4 μB, μ₄d = 1.9 μB, 2a = 0 μB)
magmoms = [0, 0,  # 2a sites (Ge)
           -3.4, -3.4,  # 2b sites (Mn)
           1.9, 1.9, 1.9, 1.9]  # 4d sites (Mn)
structure.add_site_property("magmom", magmoms)

# ── 5. CRYSTAL-TOOLKIT COMPONENT (with transparency) ────────
scene_settings = {
    "polyhedra_opacity": 0.1,
    "unit_cell_opacity": 0.05,
}
component = StructureMoleculeComponent(structure, scene_settings=scene_settings)

# ── 6. CIF and POSCAR EXPORT (rendered as <pre>) ─────────────
cif_string = structure.to(fmt="cif")
cif_display = cif_string.replace("\n", "<br>").replace(" ", " ")

poscar_string = structure.to(fmt="poscar")
poscar_display = poscar_string.replace("\n", "<br>").replace(" ", " ")

# ── 7. DASH APP LAYOUT ──────────────────────────────────────
app = Dash(__name__)
app.layout = html.Div(
    [
        html.H2(f"🧱 Heusler Unit Cell: 2b = {X_ELEMENT_2b}, "
                f"4d = {X_ELEMENT_4d}, 2a = {Z_ELEMENT}"),
        component.layout(),
        html.H2("📄 Generated CIF"),
        html.Div(
            children=cif_display,
            style={
                "whiteSpace": "pre-wrap",
                "fontFamily": "monospace",
                "backgroundColor": "#f4f4f4",
                "padding": "10px",
                "borderRadius": "5px",
            },
        ),
        html.H2("📄 Generated POSCAR"),
        html.Div(
            children=poscar_display,
            style={
                "whiteSpace": "pre-wrap",
                "fontFamily": "monospace",
                "backgroundColor": "#f4f4f4",
                "padding": "10px",
                "borderRadius": "5px",
            },
        ),
    ]
)

# ── 8. RUN ──────────────────────────────────────────────────
app.run(debug=False, port=random.randint(8051, 8999))
```

```python
from pymatgen.core import Lattice, Structure
from crystal_toolkit.components.structure import StructureMoleculeComponent
from dash import Dash, html
import random

# ── 1. USER CHOICES ─────────────────────────────────────────
X_ELEMENT_2b = "Mn"  # Mn at 2b sites
X_ELEMENT_4d = "Mn"  # Mn at 4d sites
Z_ELEMENT = "Ge"     # Ge at 2a sites

a = 4.045             # Å  ┐  tetragonal lattice parameters
c = 5.967             # Å  ┘  (set c = a if you want cubic)

# ── 2. BUILD LATTICE ────────────────────────────────────────
lattice = Lattice.from_parameters(a, a, c, 90, 90, 90)

# ── 3. ORIGINAL FRACTIONAL COORDINATES (body-centered tetragonal, 2 f.u.) ────
#    Wyckoff positions for space group I4/mmm (No. 139):
#    - 2a: (0, 0, 0), (0, 0, ½) for Ge
#    - 2b: (0, 0, ½), (0, 0, 0) for Mn (adjusted for 2b)
#    - 4d: (0, ½, ¼), (0, ½, ¾), (½, 0, ¼), (½, 0, ¾) for Mn
original_frac_coords = [
    # 2a sites (Ge)
    [0.00, 0.00, 0.00],  # Ge
    [0.50, 0.50, 0.50],  # Ge
    # 2b sites (Mn)
    [0.50, 0.50, 0.00],  # Mn
    [0.00, 0.00, 0.50],  # Mn
    # 4d sites (Mn)
    [0.00, 0.50, 0.25],  # Mn
    [0.00, 0.50, 0.75],  # Mn
    [0.50, 0.00, 0.25],  # Mn
    [0.50, 0.00, 0.75],  # Mn
]

# Coordinate shift settings (user can modify here)
dx = -0.25    # Shift along x-axis
dy = -0.25    # Shift along y-axis
dz = -0.125   # Shift along z-axis (e.g., set to 0.5 for ½ shift)

# ── 4. APPLY COORDINATE SHIFT WITH WRAPPING (mod 1) ─────────
shifted_frac_coords = [
    [(x + dx) % 1, (y + dy) % 1, (z + dz) % 1] 
    for x, y, z in original_frac_coords
]

# Species list matching the coords order
species = [
    Z_ELEMENT, Z_ELEMENT,  # 2a sites
    X_ELEMENT_2b, X_ELEMENT_2b,  # 2b sites
    *[X_ELEMENT_4d] * 4  # 4d sites
]

# ── 5. CREATE Structure with Magnetic Moments ───────────────
structure = Structure(lattice, species, shifted_frac_coords)
# Assign magnetic moments (Case 1: μ₂b = -3.4 μB, μ₄d = 1.9 μB, 2a = 0 μB)
magmoms = [0, 0,  # 2a sites (Ge)
           -3.4, -3.4,  # 2b sites (Mn)
           1.9, 1.9, 1.9, 1.9]  # 4d sites (Mn)
structure.add_site_property("magmom", magmoms)

# ── 6. CRYSTAL-TOOLKIT COMPONENT (with transparency) ────────
scene_settings = {
    "polyhedra_opacity": 0.1,
    "unit_cell_opacity": 0.05,
}
component = StructureMoleculeComponent(structure, scene_settings=scene_settings)

# ── 7. CIF and POSCAR EXPORT (rendered as <pre>) ─────────────
cif_string = structure.to(fmt="cif")
cif_display = cif_string.replace("\n", "<br>").replace(" ", " ")

poscar_string = structure.to(fmt="poscar")
poscar_display = poscar_string.replace("\n", "<br>").replace(" ", " ")

# ── 8. DASH APP LAYOUT ──────────────────────────────────────
app = Dash(__name__)
app.layout = html.Div(
    [
        html.H2(f"🧱 Heusler Unit Cell: 2b = {X_ELEMENT_2b}, "
                f"4d = {X_ELEMENT_4d}, 2a = {Z_ELEMENT} (Shifted by dx={dx}, dy={dy}, dz={dz})"),
        component.layout(),
        html.H2("📄 Generated CIF"),
        html.Div(
            children=cif_display,
            style={
                "whiteSpace": "pre-wrap",
                "fontFamily": "monospace",
                "backgroundColor": "#f4f4f4",
                "padding": "10px",
                "borderRadius": "5px",
            },
        ),
        html.H2("📄 Generated POSCAR"),
        html.Div(
            children=poscar_display,
            style={
                "whiteSpace": "pre-wrap",
                "fontFamily": "monospace",
                "backgroundColor": "#f4f4f4",
                "padding": "10px",
                "borderRadius": "5px",
            },
        ),
    ]
)

# ── 9. RUN ──────────────────────────────────────────────────
app.run(debug=False, port=random.randint(8051, 8999))
```

```python
from pymatgen.core import Lattice, Structure, Element
from crystal_toolkit.components.structure import StructureMoleculeComponent
from dash import Dash, html
import random

# ── 1. USER CHOICES ─────────────────────────────────────────
X_ELEMENT_2b = "Mn"  # Mn at 2b sites
X_ELEMENT_4d = "Mn"  # Mn at 4d sites
Z_ELEMENT = "Ge"     # Ge at 2a sites

a = 4.045            # Å  ┐  tetragonal lattice parameters
c = 5.             # Å  ┘  (set c = a if you want cubic)
b = a                 # Ensure b = c for tetragonal symmetry

# ── 2. BUILD LATTICE ────────────────────────────────────────
lattice = Lattice.from_parameters(a, a, c, 90, 90, 90)

# ── 3. ORIGINAL FRACTIONAL COORDINATES (body-centered tetragonal, 2 f.u.) ────
#    Wyckoff positions for space group I4/mmm (No. 139):
#    - 2a: (0, 0, 0), (0, 0, ½) for Ge
#    - 2b: (0, 0, ½), (0, 0, 0) for Mn (adjusted for 2b)
#    - 4d: (0, ½, ¼), (0, ½, ¾), (½, 0, ¼), (½, 0, ¾) for Mn
original_frac_coords = [
    # 2a sites (Ge)
    [0.00, 0.00, 0.00],  # Ge
    [0.50, 0.50, 0.50],  # Ge
    # 2b sites (Mn)
    [0.50, 0.50, 0.00],  # Mn
    [0.00, 0.00, 0.50],  # Mn
    # 4d sites (Mn)
    [0.00, 0.50, 0.25],  # Mn
    [0.00, 0.50, 0.75],  # Mn
    [0.50, 0.00, 0.25],  # Mn
    [0.50, 0.00, 0.75],  # Mn
]

# Coordinate shift settings (user can modify here)
dx = 0    # Shift along x-axis
dy = 0    # Shift along y-axis
dz = 0    # Shift along z-axis (e.g., set to 0.5 for ½ shift)

# ── 4. APPLY COORDINATE SHIFT WITH WRAPPING (mod 1) ─────────
shifted_frac_coords = [
    [(x + dx) % 1, (y + dy) % 1, (z + dz) % 1] 
    for x, y, z in original_frac_coords
]

# Species list matching the coords order
species = [
    Z_ELEMENT, Z_ELEMENT,  # 2a sites
    X_ELEMENT_2b, X_ELEMENT_2b,  # 2b sites
    *[X_ELEMENT_4d] * 4  # 4d sites
]

# ── 5. CREATE Structure with Magnetic Moments ───────────────
structure = Structure(lattice, species, shifted_frac_coords)
# Assign magnetic moments (0.0 for 2a, 3.0 for 2b, -2.0 for 4d)
magmoms = [0.0, 0.0,  # 2a sites (Ge)
           3.0, 3.0,   # 2b sites (Mn)
           -2.0, -2.0, -2.0, -2.0]  # 4d sites (Mn)
structure.add_site_property("magmom", magmoms)

# Extract unique species and numbering for .fdf file
unique_species = sorted(set(species))
species_numbering = {elem: idx + 1 for idx, elem in enumerate(unique_species)}
atomic_numbers = {elem: Element(elem).Z for elem in unique_species}

# ── 6. CRYSTAL-TOOLKIT COMPONENT (with transparency) ────────
scene_settings = {
    "polyhedra_opacity": 0.1,
    "unit_cell_opacity": 0.05,
}
component = StructureMoleculeComponent(structure, scene_settings=scene_settings)

# ── 7. CIF and POSCAR EXPORT (rendered as <pre>) ─────────────
cif_string = structure.to(fmt="cif")
# Append custom DM.InitSpin block to CIF
cif_string += "\n# Custom block for SIESTA compatibility\n"
cif_string += "%block DM.InitSpin\n"
for i, magmom in enumerate(magmoms, 1):
    cif_string += f"  {i} {magmom}\n"
cif_string += "%endblock DM.InitSpin\n"
cif_display = cif_string.replace("\n", "<br>").replace(" ", " ")


# ── 8. WRITE .FDF FILE ──────────────────────────────────────
system_label = "Mn3Ge"
with open(f"{system_label}.fdf", "wt") as f:
    # ── Header ──────────────────────────────────────────────
    f.write("# ---------------------------------------------------------------------------\n")
    f.write("# Name and Label\n")
    f.write("# ---------------------------------------------------------------------------\n\n")
    f.write(f"SystemName          {system_label}\n")
    f.write(f"SystemLabel         {system_label}\n\n")

    # ── Lattice Vectors ─────────────────────────────────────
    f.write("# ---------------------------------------------------------------------------\n")
    f.write("# Lattice\n")
    f.write("# ---------------------------------------------------------------------------\n\n")
    f.write("LatticeConstant             1.00 Ang\n\n")
    f.write("%block LatticeVectors\n")
    for vec in structure.lattice.matrix:  # Use structure.lattice.matrix instead of lattice_vectors
        f.write(f"    {vec[0]:.6f}    {vec[1]:.6f}    {vec[2]:.6f}\n")
    f.write("%endblock LatticeVectors\n\n")

    # ── Chemical Species Info ───────────────────────────────
    f.write("# ---------------------------------------------------------------------------\n")
    f.write("# Species and Atoms\n")
    f.write("# ---------------------------------------------------------------------------\n\n")
    f.write(f"NumberOfSpecies        {len(unique_species)}\n")
    f.write(f"NumberOfAtoms          {len(structure)}\n\n")

    f.write("%block ChemicalSpeciesLabel\n")
    for elem in unique_species:
        idx = species_numbering[elem]
        Z = atomic_numbers[elem]
        f.write(f"    {idx}    {Z}    {elem}\n")
    f.write("%endblock ChemicalSpeciesLabel\n\n")

    # ── Atomic Coordinates ──────────────────────────────────
    f.write("# ---------------------------------------------------------------------------\n")
    f.write("# Atomic Coordinates\n")
    f.write("# ---------------------------------------------------------------------------\n\n")
    f.write("AtomicCoordinatesFormat Ang\n\n")
    f.write("%block AtomicCoordinatesAndAtomicSpecies\n")
    for site in structure:
        x, y, z = site.coords
        species_id = species_numbering[site.specie.symbol]
        f.write(f"    {x:.8f}    {y:.8f}    {z:.8f}    {species_id}\n")
    f.write("%endblock AtomicCoordinatesAndAtomicSpecies\n\n")

    # ── Magnetic Moments ────────────────────────────────────
    f.write("# ---------------------------------------------------------------------------\n")
    f.write("# Initial Spin\n")
    f.write("# ---------------------------------------------------------------------------\n\n")
    f.write("%block DM.InitSpin\n")
    for i, magmom in enumerate(magmoms, 1):
        f.write(f"    {i}    {magmom}\n")
    f.write("%endblock DM.InitSpin\n\n")

    # ── Additional SIESTA Settings ──────────────────────────
    f.write("""
# ----------------------------------------------------------------------------
# K-Points Grid (Monkhorst-Pack)
# ----------------------------------------------------------------------------
%block kgrid_Monkhorst_Pack
     15  0  0  0.0    # k-points along a
     0  15  0  0.0    # k-points along b
     0  0   1  0.0    # k-points along c
%endblock kgrid_Monkhorst_Pack

# ----------------------------------------------------------------------------
# Geometry Constraints
# ----------------------------------------------------------------------------
%block Geometry.Constraints
    cellside a b                     # Constrain all cell sides
%endblock Geometry.Constraints

# ----------------------------------------------------------------------------
# Basis and Pseudopotential Settings
# ----------------------------------------------------------------------------
PAO.EnergyShift       0.01 eV
PAO.BasisSize         DZP

# ----------------------------------------------------------------------------
# Exchange-Correlation Functional
# ----------------------------------------------------------------------------
XC.functional         GGA
XC.authors            PBE

# ----------------------------------------------------------------------------
# Spin and Solution Settings
# ----------------------------------------------------------------------------
SpinPolarized         .true.
SolutionMethod        diagon
Spin                  Non-Collinear
SpinOrbit .true.
# ----------------------------------------------------------------------------
# Mesh and Temperature Settings
# ----------------------------------------------------------------------------
MeshCutoff            600 Ry
ElectronicTemperature 0.1 eV

# ----------------------------------------------------------------------------
# Density Matrix Mixing
# ----------------------------------------------------------------------------
DM.MixingWeight       0.2
DM.NumberPulay        20
DM.UseSaveDM          .true.
UseSaveData           .true.
SaveRho               .true.
SaveHS                .true.
TS.HS.Save            .true.
# ----------------------------------------------------------------------------
# Molecular Dynamics Settings
# ----------------------------------------------------------------------------
MD.TypeOfRun          Fire
MD.NumCGsteps         100         # ❗ Set to 0 steps (was 00) — review if intended!
MD.VariableCell       .true.
MD.MaxStressTol       0.5 GPa
MD.MaxForceTol        0.01 eV/Ang
MD.MaxCGDispl         0.02 Ang
MD.Forces.Stress      .true.

# ----------------------------------------------------------------------------
# Output Settings
# ----------------------------------------------------------------------------
WriteCoorXmol         .true.
WriteCoorStep         .true.
WriteForces           .true.
WriteHirshfeldPop     .true.
WriteVoronoiPop       .true.

# ----------------------------------------------------------------------------
# Parallelization
# ----------------------------------------------------------------------------
Diag.ParallelOverK    .true.
""")
print(f"✅ FDF file successfully written as '{system_label}.fdf' with {len(structure)} atoms.")

# ── 10. RUN ────────────────────────────────────────────────
app.run(debug=False, port=random.randint(8051, 8999))
```
