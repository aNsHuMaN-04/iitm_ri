## Building the Heterostructure

```python
from pymatgen.core import Structure, Lattice
from crystal_toolkit.components import StructureMoleculeComponent
from dash import Dash, html
import random

# ======================
# HETEROSTRUCTURE BUILDER
# ======================

# Parameters (adjust these as needed)
Fe_MgO_distance = 2.2        # Ångstrom gap between Fe and first MgO layer
MgO_Mn3Ge_distance = 2.95     # Ångstrom gap between first MgO and Mn3Ge
Mn3Ge_MgO_distance = 2.95     # Ångstrom gap between Mn3Ge and second MgO
a, b = 4.03, 4.03           # in-plane lattice constants

# Calculate maximum z-coordinates
max_z_fe = max(coord[2] for coord in Fe_coords_L)
max_z_mgo1 = max(coord[2] for coord in Mg_coords_1)
max_z_mn3ge = max(coord[2] for coord in Mn3Ge_coords)
max_z_mgo2 = max(coord[2] for coord in Mg_coords_2)  # Assuming Mg_coords_2 exists

# Calculate layer shifts
mgo1_z_shift = max_z_fe + Fe_MgO_distance
mn3ge_z_shift = max_z_mgo1 + mgo1_z_shift + MgO_Mn3Ge_distance
mgo2_z_shift = max_z_mn3ge + mn3ge_z_shift + Mn3Ge_MgO_distance

# Shift all layers
shifted_MgO1_coords = [[x, y, z + mgo1_z_shift] for x, y, z in Mg_coords_1]
shifted_Mn3Ge_coords = [[x, y, z + mn3ge_z_shift] for x, y, z in Mn3Ge_coords]
shifted_MgO2_coords = [[x, y, z + mgo2_z_shift] for x, y, z in Mg_coords_2]

# Combine all components
combined_species = Fe_species_L + Mg_species_1 + Mn3Ge_species + Mg_species_2
combined_coords = Fe_coords_L + shifted_MgO1_coords + shifted_Mn3Ge_coords + shifted_MgO2_coords
combined_magmoms = Fe_magmoms_L + Mg_magmoms_1 + Mn3Ge_magmoms + Mg_magmoms_2

# Calculate final cell c-axis length (add 25% of lattice constant as vacuum)
c_final = max(coord[2] for coord in combined_coords) + 2.2

# ======================
# CREATE STRUCTURE
# ======================

lattice = Lattice.from_parameters(a, b, c_final, 90, 90, 90)
structure = Structure(
    lattice=lattice,
    species=combined_species,
    coords=combined_coords,
    coords_are_cartesian=True,
    site_properties={"magmom": combined_magmoms}
)

# ======================
# DATA OUTPUT FORMATTING
# ======================

# Format coordinates with species and magmoms
output_lines = []
for species, coord, magmom in zip(combined_species, combined_coords, combined_magmoms):
    x = f"{float(coord[0]):.8f}"
    y = f"{float(coord[1]):.8f}"
    z = f"{float(coord[2]):.8f}"
    mag = f"{float(magmom):.3f}"
    line = f"{x:<12} {y:<12} {z:<12} {mag:<12} {species}"
    output_lines.append(line)

# Prepare CIF text for HTML display
cif_text = structure.to(fmt="cif")
cif_html = cif_text.replace("\n", "<br>").replace(" ", " ")

# ======================
# VISUALIZATION
# ======================

# Crystal Toolkit viewer
viewer = StructureMoleculeComponent(structure)

# Dash App layout
app = Dash(__name__)
app.layout = html.Div([
    html.H2("🧱 Fe-MgO-Mn3Ge-MgO Quadrilayer Heterostructure Viewer"),
    viewer.layout(),
    html.H2("📄 CIF of Heterostructure"),
    html.Div(children=cif_html, style={
        'whiteSpace': 'pre-wrap',
        'fontFamily': 'monospace',
        'backgroundColor': '#f4f4f4',
        'padding': '10px',
        'borderRadius': '5px',
    }),
    html.H2("📋 Formatted Coordinates with Species and Magnetic Moments"),
    html.Pre("\n".join(output_lines), style={
        'whiteSpace': 'pre-wrap',
        'fontFamily': 'monospace',
        'backgroundColor': '#f4f4f4',
        'padding': '10px',
        'borderRadius': '5px',
    })
])

# ======================
# PRINT SUMMARY
# ======================

print("="*60)
print(f"Successfully created Fe-MgO-Mn3Ge-MgO quadrilayer heterostructure")
print(f"Total atoms: {len(combined_species)}")
print(f"  - Fe atoms: {len(Fe_species_L)}")
print(f"  - First MgO atoms: {len(Mg_species_1)}")
print(f"  - Mn3Ge atoms: {len(Mn3Ge_species)}")
print(f"  - Second MgO atoms: {len(Mg_species_2)}")
print(f"Final cell dimensions: {a:.2f} × {b:.2f} × {c_final:.2f} Å")
print(f"Interlayer gaps:")
print(f"  Fe-MgO1: {Fe_MgO_distance:.2f} Å")
print(f"  MgO1-Mn3Ge: {MgO_Mn3Ge_distance:.2f} Å")
print(f"  Mn3Ge-MgO2: {Mn3Ge_MgO_distance:.2f} Å")
print("="*60)

# Run Dash app
app.run(debug=False, port=random.randint(8051, 8999))
```
