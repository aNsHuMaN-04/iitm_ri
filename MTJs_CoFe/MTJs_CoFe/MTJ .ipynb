```python
from pymatgen.core import Structure, Lattice
from crystal_toolkit.components import StructureMoleculeComponent
from dash import Dash, html
import random

# ======================
# HETEROSTRUCTURE BUILDER
# ======================

# Parameters (adjust these as needed)
Fe_Dev_distance = 2.860247*0.25  
Dev_Mn3Ge_distance = 5.9162*0.25 ### VVIP Step. Closely check the Junction
a, b = 4.045, 4.045           

# Calculate maximum z-coordinates
max_z_fe = max(coord[2] for coord in Fe_coords_L)
max_z_mgo = max(coord[2] for coord in Dev_coords)
max_z_mn3ge = max(coord[2] for coord in Mn3Ge_coords)

# Calculate layer shifts
mgo_z_shift = max_z_fe + Fe_Dev_distance
mn3ge_z_shift = max_z_mgo + mgo_z_shift + Dev_Mn3Ge_distance

# Shift layers
shifted_MgO_coords = [[x, y, z + mgo_z_shift] for x, y, z in Dev_coords]
shifted_Mn3Ge_coords = [[x, y, z + mn3ge_z_shift] for x, y, z in Mn3Ge_coords]

# Combine all components
combined_species = Fe_species_L + Dev_species + Mn3Ge_species
combined_coords = Fe_coords_L + shifted_MgO_coords + shifted_Mn3Ge_coords
combined_magmoms = Fe_magmoms_L + Dev_magmoms + Mn3Ge_magmoms

# Calculate final cell c-axis length (add 25% of lattice constant as vacuum)
c_final = max(coord[2] for coord in combined_coords) + 0.25*5.9162

# ======================
# CREATE STRUCTURE
# ======================

lattice = Lattice.from_parameters(a, b, c_final, 90, 90, 90)
structure = Structure(
    lattice=lattice,
    species=combined_species,
    coords=combined_coords,
    coords_are_cartesian=True,
    site_properties={"magmom": combined_magmoms}
)

# ======================
# DATA OUTPUT FORMATTING
# ======================

# Format coordinates with species and magmoms
output_lines = []
for species, coord, magmom in zip(combined_species, combined_coords, combined_magmoms):
    x = f"{float(coord[0]):.8f}"
    y = f"{float(coord[1]):.8f}"
    z = f"{float(coord[2]):.8f}"
    mag = f"{float(magmom):.3f}"
    line = f"{x:<12} {y:<12} {z:<12} {mag:<12} {species}"
    output_lines.append(line)

# Prepare CIF text for HTML display
cif_text = structure.to(fmt="cif")
cif_html = cif_text.replace("\n", "<br>").replace(" ", " ")

# ======================
# VISUALIZATION
# ======================

# Crystal Toolkit viewer
viewer = StructureMoleculeComponent(structure)

# Dash App layout
app = Dash(__name__)
app.layout = html.Div([
    html.H2("🧱 Fe-MgO-Mn3Ge Trilayer Heterostructure Viewer"),
    viewer.layout(),
    html.H2("📄 CIF of Heterostructure"),
    html.Div(children=cif_html, style={
        'whiteSpace': 'pre-wrap',
        'fontFamily': 'monospace',
        'backgroundColor': '#f4f4f4',
        'padding': '10px',
        'borderRadius': '5px',
    }),
    html.H2("📋 Formatted Coordinates with Species and Magnetic Moments"),
    html.Pre("\n".join(output_lines), style={
        'whiteSpace': 'pre-wrap',
        'fontFamily': 'monospace',
        'backgroundColor': '#f4f4f4',
        'padding': '10px',
        'borderRadius': '5px',
    })
])

# ======================
# PRINT SUMMARY
# ======================

print("="*60)
print(f"Successfully created Fe-MgO-Mn3Ge trilayer heterostructure")
print(f"Total atoms: {len(combined_species)}")
print(f"  - Fe atoms: {len(Fe_species_L)}")
print(f"  - MgO atoms: {len(Dev_species)}")
print(f"  - Mn3Ge atoms: {len(Mn3Ge_species)}")
print(f"Final cell dimensions: {a:.2f} × {b:.2f} × {c_final:.2f} Å")
print(f"Interlayer gaps:")
print(f"  Fe-MgO: {Fe_Dev_distance:.2f} Å")
print(f"  MgO-Mn3Ge: {Dev_Mn3Ge_distance:.2f} Å")
print("="*60)

# Run Dash app
app.run(debug=False, port=random.randint(8251, 8999))
```

```python
from pymatgen.core import Structure, Lattice
from crystal_toolkit.components import StructureMoleculeComponent
from dash import Dash, html
import random

# ======================
# HETEROSTRUCTURE BUILDER
# ======================

# Parameters (adjust these as needed)
Fe_Dev_distance = 2.860247*0.25  
Dev_Mn3Ge_distance = 5.9162*0.25  # VVIP step
a, b = 4.045, 4.045

# >>> NEW: global z-shift controls <<<
GLOBAL_Z_SHIFT =  2.860247*0.25  # Å; +ve shifts up, -ve shifts down
WRAP_IN_CELL   = False  # if True, wrap z into [0, c_final) after shifting

# Calculate maximum z-coordinates
max_z_fe    = max(coord[2] for coord in Fe_coords_L)
max_z_mgo   = max(coord[2] for coord in Dev_coords)
max_z_mn3ge = max(coord[2] for coord in Mn3Ge_coords)

# Calculate layer shifts
mgo_z_shift   = max_z_fe + Fe_Dev_distance
mn3ge_z_shift = max_z_mgo + mgo_z_shift + Dev_Mn3Ge_distance

# Shift layers
shifted_MgO_coords   = [[x, y, z + mgo_z_shift]   for x, y, z in Dev_coords]
shifted_Mn3Ge_coords = [[x, y, z + mn3ge_z_shift] for x, y, z in Mn3Ge_coords]

# Combine all components (before global shift)
combined_species = Fe_species_L + Dev_species + Mn3Ge_species
combined_coords  = Fe_coords_L + shifted_MgO_coords + shifted_Mn3Ge_coords
combined_magmoms = Fe_magmoms_L + Dev_magmoms + Mn3Ge_magmoms

# >>> apply global z-shift to all atoms <<<
combined_coords = [[x, y, z + GLOBAL_Z_SHIFT] for x, y, z in combined_coords]

# Calculate final cell c-axis length (vacuum = 25% of 5.9162 Å)
c_final = max(coord[2] for coord in combined_coords) + 0.25*5.9162 - GLOBAL_Z_SHIFT

# If desired, wrap all atoms back into the simulation cell along z
if WRAP_IN_CELL:
    combined_coords = [[x, y, (z % c_final)] for x, y, z in combined_coords]

# ======================
# CREATE STRUCTURE
# ======================

lattice = Lattice.from_parameters(a, b, c_final, 90, 90, 90)
structure = Structure(
    lattice=lattice,
    species=combined_species,
    coords=combined_coords,
    coords_are_cartesian=True,
    site_properties={"magmom": combined_magmoms}
)

# ======================
# DATA OUTPUT FORMATTING
# ======================

output_lines = []
for species, coord, magmom in zip(combined_species, combined_coords, combined_magmoms):
    x = f"{float(coord[0]):.8f}"
    y = f"{float(coord[1]):.8f}"
    z = f"{float(coord[2]):.8f}"
    # magmom may be scalar or vector; print scalar by default
    mag = magmom if isinstance(magmom, (int, float)) else (
        magmom[2] if isinstance(magmom, (list, tuple)) and len(magmom) >= 3 else magmom[0]
    )
    mag = f"{float(mag):.3f}"
    line = f"{x:<12} {y:<12} {z:<12} {mag:<12} {species}"
    output_lines.append(line)

cif_text = structure.to(fmt="cif")
cif_html = cif_text.replace("\n", "<br>").replace(" ", " ")

# ======================
# VISUALIZATION
# ======================

viewer = StructureMoleculeComponent(structure)

app = Dash(__name__)
app.layout = html.Div([
    html.H2("🧱 Fe-MgO-Mn3Ge Trilayer Heterostructure Viewer"),
    viewer.layout(),
    html.H2("📄 CIF of Heterostructure"),
    html.Div(children=cif_html, style={
        'whiteSpace': 'pre-wrap',
        'fontFamily': 'monospace',
        'backgroundColor': '#f4f4f4',
        'padding': '10px',
        'borderRadius': '5px',
    }),
    html.H2("📋 Formatted Coordinates with Species and Magnetic Moments"),
    html.Pre("\n".join(output_lines), style={
        'whiteSpace': 'pre-wrap',
        'fontFamily': 'monospace',
        'backgroundColor': '#f4f4f4',
        'padding': '10px',
        'borderRadius': '5px',
    })
])

# ======================
# PRINT SUMMARY
# ======================

print("="*60)
print("Successfully created Fe-MgO-Mn3Ge trilayer heterostructure")
print(f"Total atoms: {len(combined_species)}")
print(f"  - Fe atoms:   {len(Fe_species_L)}")
print(f"  - MgO atoms:  {len(Dev_species)}")
print(f"  - Mn3Ge atoms:{len(Mn3Ge_species)}")
print(f"Final cell dimensions: {a:.2f} × {b:.2f} × {c_final:.2f} Å")
print("Interlayer gaps:")
print(f"  Fe-MgO:     {Fe_Dev_distance:.2f} Å")
print(f"  MgO-Mn3Ge:  {Dev_Mn3Ge_distance:.2f} Å")
print(f"Global z-shift applied: {GLOBAL_Z_SHIFT:.2f} Å (WRAP_IN_CELL={WRAP_IN_CELL})")
print("="*60)

# Run Dash app
app.run(debug=False, port=random.randint(8051, 8999))
```
