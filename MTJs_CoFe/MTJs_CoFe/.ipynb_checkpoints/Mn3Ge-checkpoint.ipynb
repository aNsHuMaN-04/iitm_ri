## Mn3Ge Array Creation
The Mn3Ge super cell is created in 3 steps.

```python
from pymatgen.core.lattice import Lattice

def generate_heusler_coordinates(a, b, c, X_2b_element="Mn", X_4d_element="Mn", Z_element="Ge", 
                               dx=0.0, dy=0.0, dz=0.0, z_shift=0.0):
    """
    Generates atomic coordinates for an inverse Heusler structure with customizable elements and magnetic moments.
    
    Args:
        a, b, c: Lattice parameters (Å)
        X_2b_element: Element for 2b Wyckoff sites (default: "Mn")
        X_4d_element: Element for 4d Wyckoff sites (default: "Mn")
        Z_element: Element for 2a Wyckoff sites (default: "Ge")
        dx, dy, dz: Fractional coordinate shifts
        z_shift: Additional z-axis shift in Å (real coordinates)
        
    Returns:
        Dictionary containing:
        - 'Z_atoms': (element, coords, magmoms) for 2a sites
        - 'X_2b_atoms': (element, coords, magmoms) for 2b sites
        - 'X_4d_atoms': (element, coords, magmoms) for 4d sites
        - 'last_z_coordinate': z-coordinate of the last atom in Cartesian coordinates
    """
    # Define fractional coordinates for inverse Heusler structure (space group F-43m)
    frac_coords = [
        # 2a sites (Z atoms)
        [0.00, 0.00, 0.00],  # Z
        [0.50, 0.50, 0.50],  # Z
        # 2b sites (X_2b atoms)
        [0.50, 0.50, 0.00],  # X_2b
        [0.00, 0.00, 0.50],  # X_2b
        # 4d sites (X_4d atoms)
        [0.50, 0.00, 0.25],  # X_4d
        [0.50, 0.00, 0.75],  # X_4d
        [0.00, 0.50, 0.25],  # X_4d
        [0.00, 0.50, 0.75],  # X_4d
    ]
    
    # Assign species to each site
    species = [
        Z_element, Z_element,           # 2a sites
        X_2b_element, X_2b_element,     # 2b sites
        X_4d_element, X_4d_element,     # 4d sites
        X_4d_element, X_4d_element      # 4d sites
    ]
    
    # Assign magnetic moments
    magmoms = [
        0.0, 0.0,           # 2a sites (Z atoms, no magnetic moment)
        3.0, 3.0,           # 2b sites (X_2b atoms, magmom = 3)
        -1.0, -1.0,         # 4d sites (X_4d atoms, magmom = -1)
        -1.0, -1.0          # 4d sites (X_4d atoms, magmom = -1)
    ]
    
    # Apply fractional shifts with wrapping to [0,1)
    shifted_frac_coords = [
        [(x + dx) % 1, (y + dy) % 1, (z + dz) % 1]
        for x, y, z in frac_coords
    ]
    
    # Convert to Cartesian coordinates
    lattice = Lattice.from_parameters(a, b, c, 90, 90, 90)
    cart_coords = [lattice.get_cartesian_coords(coord) for coord in shifted_frac_coords]
    
    # Apply z_shift in Cartesian coordinates (Å)
    if z_shift != 0:
        cart_coords = [[x, y, z + z_shift] for x, y, z in cart_coords]
    
    # Group coordinates and magnetic moments by atom type and Wyckoff position
    return {
        'Z_atoms': (Z_element, cart_coords[0:2], magmoms[0:2]),        # 2a sites
        'X_2b_atoms': (X_2b_element, cart_coords[2:4], magmoms[2:4]),  # 2b sites
        'X_4d_atoms': (X_4d_element, cart_coords[4:8], magmoms[4:8]),  # 4d sites
        'last_z_coordinate': cart_coords[-1][2]                        # Last z-coordinate
    }
```

### Provide the number of layers. 

```python
# ===================================== Enter the Number of Layers ===========================================
n_layers = 36  # Example input
# ============================================================================================================
```

```python
from pymatgen.core.lattice import Lattice
import math

layers_per_unit_cell = 4
n_cells = math.ceil(n_layers / layers_per_unit_cell)

# Lattice parameters
a, b, c = 4.045, 4.045, 5.9162

# Generate n cascaded groups
groups = []
for i in range(n_cells):
    z_shift = 5.9162 * i
    group = generate_heusler_coordinates(a, b, c, dx=0.0, dy=0.0, dz=0, z_shift=z_shift)
    groups.append(group)

# Combine species, coordinates, and magnetic moments from all groups
Mn3Ge_species = []
Mn3Ge_coords = []
Mn3Ge_magmoms = []

for group in groups:
    # Add Z atoms (2 positions)
    Mn3Ge_species.extend([group['Z_atoms'][0]] * 2)
    Mn3Ge_coords.extend(group['Z_atoms'][1])
    Mn3Ge_magmoms.extend(group['Z_atoms'][2])  # Magnetic moments for 2a sites
    
    # Add X_2b atoms (2 positions)
    Mn3Ge_species.extend([group['X_2b_atoms'][0]] * 2)
    Mn3Ge_coords.extend(group['X_2b_atoms'][1])
    Mn3Ge_magmoms.extend(group['X_2b_atoms'][2])  # Magnetic moments for 2b sites
    
    # Add X_4d atoms (4 positions)
    Mn3Ge_species.extend([group['X_4d_atoms'][0]] * 4)
    Mn3Ge_coords.extend(group['X_4d_atoms'][1])
    Mn3Ge_magmoms.extend(group['X_4d_atoms'][2])  # Magnetic moments for 4d sites

# Print the results
print(f"combined_species (n={n_cells}) = {Mn3Ge_species}")
print(f"\ncombined_coords (n={n_cells}) = [")
for coord in Mn3Ge_coords:
    print(f"    {coord},")
print("]")
print(f"\ncombined_magmoms (n={n_cells}) = {Mn3Ge_magmoms}")
```

```python
from pymatgen.core import Structure, Lattice
import numpy as np
import math  # Added math import

class AtomData:
    def __init__(self, x: float, y: float, z: float, magmom: float, species: str):
        self.x = x
        self.y = y
        self.z = z
        self.magmom = magmom
        self.species = species
    
    def __repr__(self):
        return f"AtomData(x={self.x:.8f}, y={self.y:.8f}, z={self.z:.8f}, magmom={self.magmom:.3f}, species='{self.species}')"

class AtomicStructure:
    def __init__(self):
        self.atoms = []
    
    def add_atom(self, x: float, y: float, z: float, magmom: float, species: str):
        self.atoms.append(AtomData(x, y, z, magmom, species))
    
    def process_data(self, species_list, coord_list, magmom_list):
        for species, coord, magmom in zip(species_list, coord_list, magmom_list):
            x = float(coord[0])
            y = float(coord[1])
            z = float(coord[2])
            mag = float(magmom)
            self.add_atom(x, y, z, mag, species)
        
        # Sort atoms by z-coordinate
        self.atoms.sort(key=lambda atom: atom.z)
    
    def remove_last_n_atoms(self, n):
        """Remove the last n atoms (with highest z-coordinates)"""
        if len(self.atoms) >= n:
            self.atoms = self.atoms[:-n]
    
    def print_formatted(self):
        print("Index   X           Y           Z           Magmom      Species")
        print("-------------------------------------------------------------")
        for i, atom in enumerate(self.atoms, 1):
            print(f"{i:<6} {atom.x:<12.8f} {atom.y:<12.8f} {atom.z:<12.8f} {atom.magmom:<12.3f} {atom.species}")
    
    def create_pymatgen_structure(self, a: float, b: float, c: float):
        # Create lattice - now using max z-coordinate of remaining atoms
        if self.atoms:
            max_z = max(atom.z for atom in self.atoms)
            c = max_z + 0.25 * 5.9162  # Buffer for visualization
        else:
            c = c  # Default value if no atoms
        
        lattice = Lattice.from_parameters(a, b, c, 90, 90, 90)
        
        # Prepare data for Structure
        species_list = [atom.species for atom in self.atoms]
        coords_list = [[atom.x, atom.y, atom.z] for atom in self.atoms]
        magmoms_list = [atom.magmom for atom in self.atoms]
        
        # Create Structure
        structure = Structure(
            lattice,
            species_list,
            coords_list,
            coords_are_cartesian=True,
            site_properties={"magmom": magmoms_list}
        )
        
        return lattice, structure

# Example usage
structure_data = AtomicStructure()
structure_data.process_data(Mn3Ge_species, Mn3Ge_coords, Mn3Ge_magmoms)

print("Original structure:")
structure_data.print_formatted()

# Remove atoms only if n_layers is not a multiple of 4
if n_layers % 4 != 0:
    atoms_to_remove = 2 * (math.ceil(n_layers / layers_per_unit_cell) * 4 - n_layers)
    structure_data.remove_last_n_atoms(atoms_to_remove)
    print(f"\nStructure after removing last {atoms_to_remove} atoms:")
    structure_data.print_formatted()
else:
    print("\nNo atoms removed (n_layers is a multiple of 4)")

# Create pymatgen Structure with remaining atoms
lattice, structure = structure_data.create_pymatgen_structure(a, b, c)

# Display lattice information
print("\nLattice parameters:")
print(f"a = {lattice.a:.6f} Å")
print(f"b = {lattice.b:.6f} Å")
print(f"c = {lattice.c:.6f} Å")
print(f"Volume = {lattice.volume:.6f} Å³")
print("\nLattice matrix:")
print(lattice.matrix)

# Visualize the structure (if crystal-toolkit is available)
try:
    from crystal_toolkit.components import StructureMoleculeComponent
    from dash import html, Dash
    import random
    
    scene_settings = {
        "polyhedra_opacity": 0.1,
        "unit_cell_opacity": 0.05,
    }
    component = StructureMoleculeComponent(structure, scene_settings=scene_settings)
    
    app = Dash(__name__)
    app.layout = html.Div([
        html.H2("Crystal Structure Visualization"),
        component.layout()
    ])
    
    app.run(debug=False, port=random.randint(8051, 8999))
except ImportError:
    print("\nNote: crystal-toolkit not available for visualization")
```

```python
# Generate .fdf file from the structure
system_label = "Mn3Ge"

# Get unique species and assign specific numbers
unique_species = sorted({atom.species for atom in structure_data.atoms})
species_numbering = {'Ge': 2, 'Mn': 3}  # Explicit numbering as requested

# Get atomic numbers (assuming standard elements)
atomic_numbers = {
    'Mn': 25,
    'Ge': 32,
    # Add other elements if needed
}

# Write the .fdf file
with open(f"{system_label}.fdf", "w") as f:
    # ── Header ──────────────────────────────────────────────
    f.write("# SIESTA input file generated from structure data\n")
    f.write(f"SystemName          {system_label}\n")
    f.write(f"SystemLabel         {system_label}\n\n")
    
    # ── Lattice Vectors ────────────────────────────────────
    f.write("LatticeConstant             1.00 Ang\n\n")
    f.write("%block LatticeVectors\n")
    for vec in lattice.matrix:
        f.write(f"    {vec[0]:.10f}    {vec[1]:.10f}    {vec[2]:.10f}\n")
    f.write("%endblock LatticeVectors\n\n")
    
    # ── Chemical Species ───────────────────────────────────
    f.write(f"NumberOfSpecies        {len(unique_species)}\n")
    f.write(f"NumberOfAtoms          {len(structure_data.atoms)}\n\n")
    
    f.write("%block ChemicalSpeciesLabel\n")
    # Write in order: Ge first (1), then Mn (5)
    f.write(f"    1    {atomic_numbers['Ge']}    Ge\n")
    f.write(f"    5    {atomic_numbers['Mn']}    Mn\n")
    f.write("%endblock ChemicalSpeciesLabel\n\n")
    
    # ── Atomic Coordinates ─────────────────────────────────
    f.write("AtomicCoordinatesFormat Ang\n")
    f.write("%block AtomicCoordinatesAndAtomicSpecies\n")
    for atom in structure_data.atoms:
        species_num = species_numbering[atom.species]
        f.write(f"    {atom.x:.10f}    {atom.y:.10f}    {atom.z:.10f}    {species_num}\n")
    f.write("%endblock AtomicCoordinatesAndAtomicSpecies\n\n")
    
    # ── Magnetic Moments ────────────────────────────────────
    f.write("%block DM.InitSpin\n")
    for i, atom in enumerate(structure_data.atoms, 1):
        f.write(f"    {i}    {atom.magmom:.6f}\n")
    f.write("%endblock DM.InitSpin\n\n")
    
    # ── Additional SIESTA Settings ──────────────────────────
    f.write("""
%block kgrid_Monkhorst_Pack
8    0   0   0.0
0    8   0   0.0
0    0   60   0.0
%endblock kgrid_Monkhorst_Pack

==================================================
==================================================
# SCF variables

DM.MixSCF1   T
MaxSCFIterations      300           # Maximum number of SCF iter
DM.MixingWeight       0.3          # New DM amount for next SCF cycle
DM.Tolerance          1.d-4         # Tolerance in maximum difference
DM.UseSaveDM          true          # to use continuation files
DM.NumberPulay        15

==================================================
==================================================
# MD variables

MD.FinalTimeStep 1
MD.TypeOfRun CG
MD.NumCGsteps     000
MD.UseSaveXV      .true.

==================================================
==================================================
# Output variables

WriteMullikenPop                0
WriteBands                      .false.
SaveRho                         .false.
SaveDeltaRho                    .false.
SaveHS                          .false.
SaveElectrostaticPotential      True 
SaveTotalPotential              no
WriteCoorXmol                   .true.
WriteMDXmol                     .true.
WriteMDhistory                  .false.
WriteEigenvalues                no
SolutionMethod diagon
ElectronicTemperature  1200 K
SaveRho        .true.

""")

print(f"Successfully created {system_label}.fdf file with {len(structure_data.atoms)} atoms.")
```
